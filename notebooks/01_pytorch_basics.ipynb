## 4. Softmax by hand

```python
row = torch.tensor([1.0, 2.0, 3.0])

def my_softmax(v):
    e = torch.exp(v)
    return e/e.sum()
```

```python
my_softmax(row)
```

```python
torch.softmax(row, dim=-1)
```

```python
torch.allclose(my_softmax(row), torch.softmax(row, dim=-1))
```

```python
big = torch.tensor([1.0, 2.0, 1000.0])
my_softmax(big)
```

```python
torch.softmax(big, dim=-1)
```

```python
def stable_softmax(v):
    e = torch.exp(v - v.max())
    return e / e.sum()

stable_softmax(big)
```

```python
stable_softmax(row), my_softmax(row)
```

```python
print(info.max)
print(math.log(info.max))    # 88.72
print(torch.exp(torch.tensor(88)))   # 1.65e+38
print(torch.exp(torch.tensor(89.0)))   # inf
```

```python
mx = torch.tensor(info.max)
```

```python
print(mx)                    # 3.4028e+38
print(mx + 1)                # unchanged: 1 is far below the gap between floats here
print(mx + 2) 
print(mx == (mx + 1))
print(mx * 2)                # inf, saturates, does not wrap to negative
print(mx * 2 - mx * 2)       # nan: inf - inf has no answer
print(-mx * 2)               # -inf
print(torch.tensor(1.0) / torch.tensor(0.0))   # inf, not an error
```

```python
# Going small: underflow fades to zero instead of wrapping.
tiny = torch.tensor(info.smallest_normal)
print(tiny)                  # 1.1755e-38
print(tiny / 2)              # 5.8775e-39, a "denormal", still nonzero but losing digits
print(tiny / 2**24)          # 0., underflow
print(torch.exp(torch.tensor(-104.0)))   # 0.: exp(x) for x below about -103 is zero
```

```python
# Precision: the gap between neighbouring floats grows with magnitude.
print(torch.tensor(1.0) + info.eps)      # 1.0000001, the next float after 1
print(torch.tensor(1.0) + info.eps / 2)  # 1., gap too small to register
print(torch.tensor(2**24, dtype=torch.float32) + 1)   # 16777216., lost
print(torch.tensor(2**24, dtype=torch.float32) + 2)   # 16777218., gap here is 2
```

```python
# The gap at any magnitude is eps times the magnitude.
for v in [1.0, 1000.0, 1e6, 1e10]:
    t = torch.tensor(v)
    print(v, "next float is", torch.nextafter(t, torch.tensor(float("inf"))).item() - v)
```

```python
print(torch.tensor(16777216, dtype=torch.int32) + 1)       # 16777217, exact
print(torch.tensor(16777216.0, dtype=torch.float32) + 1)   # 16777216., the 1 is lost
```

## 6. nn.Module

### A tiny training loop

```python
import torch.nn as nn
```

```python
# ---- Data: y = 3x + 2 plus noise. The model must discover the 3 and the 2.
torch.manual_seed(0)
X = torch.rand(100, 1) * 10                 # 100 samples, 1 feature
y = 3 * X + 2 + torch.randn(100, 1) * 0.5   # noisy targets
```

```python
print(X.shape)
X[:5]
```

```python
print(y.shape)
y[:5]
```

```python
# ---- Model: one Linear layer, 1 input -> 1 output. That is y = w*x + b.
model = nn.Linear(1, 1)
print("before training:", model.weight.item(), model.bias.item())   # random
```

```python
model
```

```python
# ---- Loss: mean squared error, same as in linear regression.
loss_fn = nn.MSELoss()
```

```python
# ---- Optimizer: it holds a reference to the model's parameters and
# knows how to nudge them. lr is the step size.
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
optimizer
```

```python
# ---- The loop. Every PyTorch training loop has exactly these five lines.
for step in range(400):
    pred = model(X)                 # 1. forward: compute predictions
    loss = loss_fn(pred, y)         # 2. how wrong are they, as one number
    optimizer.zero_grad()           # 3. clear gradients from the last step
    loss.backward()                 # 4. autograd: fill .grad on every parameter
    optimizer.step()                # 5. move each parameter against its gradient
    if step % 40 == 0:
        print(f"step {step:3d}  loss {loss.item():.4f}  "
              f"w {model.weight.item():.3f}  b {model.bias.item():.3f}")

print("after training: ", model.weight.item(), model.bias.item())   # near 3 and 2
```

```python
# Replace the model with your own Module
class LineModel(nn.Module):
    def __init__(self):
        super().__init__()
        # nn.Parameter marks a tensor as learnable. Without it, this would be
        # a plain attribute and the optimizer would never see it.
        self.w = nn.Parameter(torch.randn(1))
        self.b = nn.Parameter(torch.zeros(1))

    def forward(self, x):
        return self.w * x + self.b

model = LineModel()
model
```

```python
print(list(model.named_parameters()))   # w and b, found automatically
```

```python
model = LineModel()
print("before training:", model.w.item(), model.b.item())   # random

loss_fn = nn.MSELoss()

optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

for step in range(400):
    pred = model(X) 
    loss = loss_fn(pred, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if step % 40 == 0:
        print(f"step {step:3d}  loss {loss.item():.4f}  "
              f"w {model.w.item():.3f}  b {model.b.item():.3f}")

print("after training: ", model.w.item(), model.b.item())   # near 3 and 2
```

```python
## Another module
import torch.nn as nn

class Tiny(nn.Module):
    def __init__(self):
        super().__init__()                # required: sets up parameter tracking
        self.fc = nn.Linear(4, 2)         # y = x @ W.T + b

    def forward(self, x):                 # called when you do model(x)
        return self.fc(x)
```

```python
model = Tiny()
model
```

```python
for name, p in model.named_parameters():
    print(name, p.shape)
```

```python
x = torch.randn(3, 4)       # 3 samples, 4 features
print(model(x).shape)
model(x)
```

```python
total = sum(p.numel() for p in model.parameters())
print(total)                # 10 = 4*2 weights + 2 biases
```

## 7. Autograd

```python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)

y = (x ** 2).sum()    # scalar; backward needs a scalar
y.backward()          # fills x.grad with dy/dx

print(x.grad)         # tensor([2., 4., 6.]) which is 2*x
print(x.grad == 2 * x)

# Gradients accumulate. Run backward again without clearing and x.grad doubles.
# This is why the training loop calls optimizer.zero_grad() every step.
y2 = (x ** 2).sum()
y2.backward()
print(x.grad)         # tensor([ 4.,  8., 12.])
x.grad.zero_()
print(x.grad)         # back to zeros
```

## 8. Devices

Move a tensor and a module to the MPS device, run a forward pass there, and move the
result back to CPU.

Then run a forward pass with the tensor on MPS and the module on CPU. Read the error;
you will meet it again.

```python
device = torch.device("mps")

model = Tiny().to(device)             # moves all parameters to the GPU
x = torch.randn(3, 4).to(device)      # moves the data too

out = model(x)
print(out.device)                     # mps:0

out_cpu = out.cpu()                   # back to CPU, e.g. for numpy or printing
print(out_cpu.device)                 # cpu

# The classic mistake: data on one device, model on another.
cpu_model = Tiny()
try:
    cpu_model(x)
except RuntimeError as e:
    print("error:", e)
# Whenever you see "Expected all tensors to be on the same device",
# something was not moved. Check the model, the inputs, and the targets.
```
